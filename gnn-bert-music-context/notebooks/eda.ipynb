# EDA — GNN–BERT Music Context

Quick exploratory checks on splits and sample graphs.

```python
from pathlib import Path
import json
import sys
ROOT = Path('..').resolve()
sys.path.insert(0, str(ROOT))
from src.utils import load_config, load_json
cfg = load_config(ROOT / 'config.yaml')
splits = Path(cfg['paths']['splits'])
for name in ['fma_small.json', 'musiccaps.json', 'deam.json']:
    p = splits / name
    if p.exists():
        d = load_json(p)
        print(name, {k: (len(v) if isinstance(v, list) else v) for k, v in d.items() if k in ('train','val','test','vocab','genres','n_downloaded','n_total','status')})
```

```python
from pathlib import Path
ex = Path(cfg['paths']['examples'])
print('example graphs:', len(list(ex.glob('*.pt'))))
print('plots:', list(Path(cfg['paths']['plots']).glob('*')))
```
